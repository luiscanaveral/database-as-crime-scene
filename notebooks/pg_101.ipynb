## Hola

| Schema | Type | Purpose |
| :--- | :--- | :--- |
| **`public`** | System/Default | The default namespace for user objects; used if no schema is specified. |
| **`pg_catalog`** | System | Stores internal system tables, built-in data types, and core functions. |
| **`information_schema`** | System | SQL-standard views for querying metadata (tables, columns, permissions). |
| **`pg_toast`** | System | Stores "Oversized-Attribute" data that exceeds the standard page size. |
| **`pg_temp_n`** | System | Managed automatically for session-specific temporary tables. |
| **`user_defined`** | Custom | Used for logical grouping (e.g., `hr`, `sales`) or multi-tenant isolation. |
