## Scenario

We want to store information about Students, the Courses they take, their Professors, and the Office Room of those professors.

### ❌ Unnormalized Form (UNF)

This raw table contains multiple values in a single cell, which breaks database rules.




1️⃣ First Normal Form (1NF / N1)
Rule: 
- Cells must contain atomic (single) values. No repeating groups or comma-separated lists.

To fix this, we split the multiple values into individual rows.

| StudentID (PK) | StudentName | Courses(PK)  | Professor | ProfessorOffice |
| --------- | ----------- | ------------- | --------- | --------------- |
| 10        | Alice       | Math | Dr. Smith | Room 401        |
| 10        | Alice       | Physics | Dr. Smith | Room 401        |
| 20        | Bob         | Chemistry     | Dr. Jones | Room 502        |

- Current Key: A composite primary key of (StudentID + Course) is needed to identify a unique row.The Problem: We have 
- Partial Dependencies. StudentName only relies on StudentID, not the Course. This violates 2NF.

2️⃣ Second Normal Form (2NF / N2)
Rule: 
- Must be in 1NF, and all non-key columns must depend on the whole primary key (no partial dependencies).
To fix this, we split the data into two tables so everything depends entirely on its respective key.

Table A: Student_Registration (Links students to courses)The composite key (StudentID + Course) is perfect here.

Summary of the Transformation

- UNF to 1NF: Flattened the table so every cell has exactly one value.
- 1NF to 2NF: Removed fields that only depended on part of the key into a new table.
- 2NF to 3NF: Removed fields that depended on other non-key fields into a new table.
